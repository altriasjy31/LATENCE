```python
import pathlib
import pathlib as P
import functools as ft
import itertools as it
import collections as clt
```

```python
import Bio.SeqIO as sio
import Bio.SeqIO.FastaIO as fsio
import Bio.Seq as sq
import Bio.SeqRecord as sr
```

```python
import pandas as pd
```

```python
swpath = pathlib.Path("/home/dataset-assist-0/datafile/shaojiangyi/latence-dataset/swissprot_2022_04.pkl")
swdf = pd.read_pickle(swpath)
```

```python
no_anno_df = swdf.loc[swdf["prop_annotations"].apply(lambda x: x == [])]
```

```python
no_anno_df
```

```python
seq_recs = (sr.SeqRecord(id=seq_name,name=seq_name,description="",seq=sq.Seq(seq_str)) for seq_name, seq_str in zip(no_anno_df["proteins"], no_anno_df["sequences"]))
```

```python
saving_path = "/home/dataset-assist-0/datafile/shaojiangyi/latence-dataset/swissprot_2022_04_no_anno.fa"
with open(saving_path, "w") as h:
    writer = fsio.FastaWriter(h, wrap=10000000)
    writer.write_records(seq_recs)
```
