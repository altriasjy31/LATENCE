```python
name_list = sw_df.loc[sw_df["prop_annotations"].apply(lambda x: x == []), "proteins"].tolist()
```

```python
data_path = ROOT / "data" / "dataset_state_ind_dict.pkl"
```

```python
with open(data_path, "rb") as h:
    data = pickle.load(h)
```

```python
data.keys()
```

```python
data["exp_train"] = defaultdict(dict)
```

```python
TASKS = [
    "cellular_component",
    "molecular_function",
    "biological_process",
]
```

```python
for t in TASKS:
    data["exp_train"][t] = {"proteins": name_list}
```

```python
saving_path = ROOT / "data" / "dataset_with_exp_train.pkl"
```

```python
with open(saving_path, "wb") as h:
    pickle.dump(data, h)
```
